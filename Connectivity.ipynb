```python
import os
from google.colab import userdata
from openai import OpenAI

try:
    # Read Groq API key from Colab Secrets
    api_key = userdata.get("GROQ_API_KEY")

    if not api_key:
        raise ValueError("GROQ_API_KEY not found in Colab Secrets.")

    print("✅ Environment ready. Sending prompt...")

    # Create Groq client
    client = OpenAI(
        api_key=api_key,
        base_url="https://api.groq.com/openai/v1"
    )

    prompt = "What is prompt engineering? Answer in one sentence."

    print("\nPrompt:")
    print(prompt)

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "user", "content": prompt}
        ]
    )

    answer = response.choices[0].message.content

    print("\nResponse:")
    print(answer)

    # Optional connectivity test
    second_prompt = "Explain AI in one sentence."

    response2 = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "user", "content": second_prompt}
        ]
    )

    print("\nConnectivity Test:")
    print(response2.choices[0].message.content)

except ValueError as e:
    print(f"❌ Error: {e}")

except Exception as e:
    print(f"❌ API/Network Error: {e}")
```
