```python
# Install (Run once in Colab)
!pip install -q groq

from google.colab import userdata
from groq import Groq

# ----------------------------
# Load API Key
# ----------------------------
try:
    api_key = userdata.get("GROQ_API_KEY")

    if not api_key:
        raise ValueError("GROQ_API_KEY not found in Colab Secrets.")

    client = Groq(api_key=api_key)
    print("✅ Connected to Groq Successfully!")

except Exception as e:
    print(f"❌ Error: {e}")
    raise


# ----------------------------
# Function to Send Prompt
# ----------------------------
def run_prompt(prompt):
    try:
        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            temperature=0.3,
            max_tokens=200,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        )

        return response.choices[0].message.content

    except Exception as e:
        return f"API Error: {e}"


# ----------------------------
# Prompt Cases
# ----------------------------
cases = [

{
"title":"CASE 1 : Hallucination / Factual Error",

"original":
"Who won the FIFA World Cup in 1942?",

"failure":
"Hallucination / Factual Error",

"cause":
"The prompt asks about an event that never happened. The model may invent an answer.",

"corrected":
"""The FIFA World Cup was not held in 1942 because of World War II.
Without making up facts, tell me which country won the previous FIFA World Cup in 1938."""
},

{
"title":"CASE 2 : Instruction Ignoring",

"original":
"List the three largest oceans on Earth. Respond using exactly three bullet points, with each bullet point on a new line starting with a dash.",

"failure":
"Instruction Ignoring",

"cause":
"The prompt may not strongly enforce the formatting rules.",

"corrected":
"""List only the three largest oceans on Earth.

Requirements:
- Exactly three bullet points.
- Each bullet starts with '-'
- One bullet per line.
- No numbering.
- No extra explanation."""
},

{
"title":"CASE 3 : Over-Refusal",

"original":
"Tell me a fun fact about the Roman Empire that is appropriate for school children.",

"failure":
"Over-Refusal / Safety Misfire",

"cause":
"Some models may incorrectly refuse harmless educational prompts because of broad safety filtering.",

"corrected":
"""Provide one fun and educational fact about the Roman Empire suitable for children aged 10-12.
Keep the answer under 50 words.
Do not include unnecessary warnings."""
}

]


# ----------------------------
# Execute Experiment
# ----------------------------

for case in cases:

    print("="*80)
    print(case["title"])
    print("="*80)

    # Original Prompt
    print("\nOriginal Prompt:")
    print(case["original"])

    original_response = run_prompt(case["original"])

    print("\nOriginal Response:")
    print(original_response)

    # Diagnosis
    print("\nDiagnosis")
    print("-"*40)

    print("Observed Failure:")
    print(case["failure"])

    print("\nRoot Cause:")
    print(case["cause"])

    print("\nAnalysis:")
    print("The original prompt may allow the model to generate an incorrect, poorly formatted, or overly cautious response because it lacks sufficient guidance.")

    # Corrected Prompt
    print("\nCorrected Prompt:")
    print(case["corrected"])

    corrected_response = run_prompt(case["corrected"])

    print("\nImproved Response:")
    print(corrected_response)

    # Verification
    print("\nVerification:")
    print("✔ The corrected prompt provides clearer instructions and should produce a more accurate and reliable response.")

    print("\n")


# ----------------------------
# Final Summary
# ----------------------------

print("="*80)
print("LESSONS LEARNED")
print("="*80)

print("1. Clear prompts reduce hallucinations.")
print("2. Explicit formatting instructions improve compliance.")
print("3. Adding context helps the model understand the task better.")
print("4. Prompt refinement improves reliability and response quality.")
print("5. Testing and diagnosing failures is an important part of prompt engineering.")
```
