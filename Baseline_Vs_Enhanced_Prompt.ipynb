```python
from google.colab import userdata
from openai import OpenAI

try:
    api_key = userdata.get("GROQ_API_KEY")

    if not api_key:
        raise ValueError("GROQ_API_KEY not found in Colab Secrets.")

    client = OpenAI(
        api_key=api_key,
        base_url="https://api.groq.com/openai/v1"
    )

    print("✅ Connected successfully!")

except Exception as e:
    print(f"❌ Error: {e}")
```

```python
# Same model and parameters for fair comparison

MODEL = "llama-3.3-70b-versatile"

TEMPERATURE = 0.7

MAX_TOKENS = 150


def send_prompt(prompt):
    response = client.chat.completions.create(
        model=MODEL,
        temperature=TEMPERATURE,
        max_tokens=MAX_TOKENS,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content
```

```python
baseline_prompt = (
    "Explain recursion."
)

enhanced_prompt = (
    "You are a computer science professor. "
    "Explain recursion to a beginner using a simple real-world analogy. "
    "Write exactly two paragraphs."
)
```

```python
baseline_response = send_prompt(baseline_prompt)

enhanced_response = send_prompt(enhanced_prompt)
```

```python
print("="*70)
print("BASELINE PROMPT")
print("="*70)

print("\nPrompt:")
print(baseline_prompt)

print("\nResponse:")
print(baseline_response)

print("\n\n")

print("="*70)
print("ENHANCED PROMPT")
print("="*70)

print("\nPrompt:")
print(enhanced_prompt)

print("\nResponse:")
print(enhanced_response)
```

```python
print("="*70)
print("COMPARISON")
print("="*70)

print(f"Baseline Length : {len(baseline_response.split())} words")

print(f"Enhanced Length : {len(enhanced_response.split())} words")

print("\nAnalysis:")

print("- Baseline prompt is short and gives a general explanation.")

print("- Enhanced prompt specifies a role, audience, analogy, and output format.")

print("- The enhanced response is usually clearer, easier to understand, and better structured.")

print("- Prompt engineering improves response quality by providing context and constraints.")
```
